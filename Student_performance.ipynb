```python
# Drop rows with missing values
student_data = student_data.dropna(axis=0)
```

```python
# Define the target variable to predict and the features to use for prediction
Y = student_data["G3"]
```

```python
# Define the features to use for prediction
# Select the input features that may influence student academic performance
#Feature	Meaning
#age- Student's age
#Medu- Mother's education level
#Fedu- Father's education level
#traveltime- Travel-time category
#studytime- Weekly study-time category
#failures- Previous class failures
#absences- Number of school absences
#G2- Second-period grade

student_features = ["age","Medu", "Fedu", "traveltime", "studytime", "failures", "absences","G2"]
```

```python
# Define the feature matrix
X = student_data[student_features]
```

```python
# Import the train_test_split function from Scikit-learn
from sklearn.model_selection import train_test_split
```

```python
# Split the dataset into 75% training data and 25% validation data
# The training data is used to teach the model
# The validation data is used to evaluate its predictive performance
# random_state ensures that the split is reproducible

train_X, val_X, train_y, val_Y = train_test_split(X, Y,random_state=0)
```

```python
# Import the Decision Tree Regressor algorithm
# A Decision Tree learns relationships between student features
# and their final academic grades
from sklearn.tree import DecisionTreeRegressor
```

```python
# Initialize the Decision Tree model
student_model = DecisionTreeRegressor(random_state=1)
```

```python
# Train the model using the training dataset
student_model.fit(X, Y)
```

```python
# Use the trained Decision Tree model to predict student grades
val_predictions = student_model.predict(val_X)

# Display the first ten predicted grades
print(val_predictions[:10])
```

```python
# Import Mean Absolute Error (MAE) to evaluate model performance
from sklearn.metrics import mean_absolute_error
```

```python
# Calculate the validation MAE
val_mae = mean_absolute_error(val_Y,val_predictions)

# Display the model's prediction error
print("Validation MAE:", val_mae)
```

```python
# Create a function to evaluate different Decision Tree sizes
# The goal is to find a tree size that produces a lower validation MAE
 # Create a Decision Tree with the specified maximum leaf nodes
def get_mae(max_leaf_nodes, train_X, val_X, train_Y, val_Y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes,random_state=0)
    
     # Train the model using the training data
    model.fit(train_X, train_Y)

    # Predict final grades using the validation data
    predictions = model.predict(val_X)

    # Calculate and return the validation MAE
    return mean_absolute_error(val_Y, predictions)
```

```python
# Test different maximum leaf-node values
# Compare the MAE scores to identify the best-performing tree size
for max_leaf_nodes in [5, 10, 20, 50, 100]:
    my_mae = get_mae(max_leaf_nodes, train_X,val_X,train_Y,val_Y)

    print("Max leaf nodes:", max_leaf_nodes, "MAE:", my_mae)
```

```python
# Import the Random Forest Regressor algorithm
# Random Forest combines multiple Decision Trees to make predictions
from sklearn.ensemble import RandomForestRegressor
```

```python

# Initialize the Random Forest model using 100 Decision Trees
forest_model = RandomForestRegressor(random_state=1)
```

```python
forest_model
```

```python
# Train the Random Forest model using the training dataset
forest_model.fit(train_X, train_Y)
```

```python
# Predict student grades using the validation dataset
forest_predictions = forest_model.predict(val_X)
```

```python
# Calculate the Mean Absolute Error of the Random Forest model
forest_mae = mean_absolute_error(val_Y, forest_predictions)
```

```python
print("Random Forest MAE:", forest_mae)
```

```python
# Compare the prediction errors of the Decision Tree and Random Forest models
# The model with the lower MAE performs better on this validation dataset
print("Decision Tree MAE:", val_mae)
print("Random Forest MAE:", forest_mae)

if forest_mae < val_mae:
    print("Random Forest performs better.")
else:
    print("Decision Tree performs better or equally.")
```
